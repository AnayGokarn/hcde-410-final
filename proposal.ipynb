## 1. Overview, Research Questions & Hypotheses

### 1.1 Overview
Housing is one of the strongest social determinants of health. Affordable, stable and well-maintained housing can lower stress, reduce exposure to environmental hazards (damp, mould, pests, overcrowding) and free up household income for food, transport and healthcare [1–3]. Cities provide that housing in very different ways:

- **Amsterdam**: a large share of homes are owned by non-profit *woningcorporaties* (housing associations). They are mixed into ordinary neighbourhoods, which are dense, walkable and built around cycling.
- **New York City**: public housing is concentrated in large NYCHA tower developments, many of them ageing and in need of repairs. Since 2016 a growing number of units have also been moved to private management through the RAD/PACT programme [17].
- **London**: council and housing-association homes, ranging from post-war estates to recent mixed-tenure schemes, with a long history of stock loss through Right to Buy.

This contrast makes it possible to ask not only *whether* neighbourhoods with more public housing have different health, but *what kind* of public housing environment goes with better or worse health.

The study is **observational and ecological**: the unit of analysis is the neighbourhood, not the person. The aim is to *describe and compare associations* carefully, not to estimate the causal effect of living in public housing. Section 2 explains why this distinction matters a lot for this topic.

### 1.3 Hypotheses
In A4 I had one hypothesis ("areas with more public housing will have better health"). Since then, my background reading has shown that the raw association could go the other way. Since social housing is allocated by need, tenants are on average likely to be poorer, older, and more likely to be disabled [7, 8]. So now I have two hypotheses for before and after adjusting for these factors.

- **Raw Hypothesis:** In all three cities, neighbourhoods with a higher social-housing share will have *worse* average health on every indicator. I expect this because of who social housing is allocated to (selection), not because of an effect of the housing.
- **Adjusted Hypothesis:** After controlling for income and demographic composition, the association between social housing and poor health outcomes will shrink and social housing will have a null or slightly favourable association with health outcomes. This would be false if the unfavourable association remains after adjustment in all three cities.
- **H2:** Changes in social-housing share over the study window will be only weakly associated with changes in health. Housing stock changes slowly and the windows are short, so I expect small or null coefficients. A clear relationship would be a notable finding.
- **H3:** The adjusted association will be more favourable when social housing is dispersed and mixed than when it is spatially concentrated. Housing with walking & cycling infrastructure will have a stronger association with better physical health. I expect Amsterdam to be the most associated with positive health outcomes.

### 1.4 Why it's useful
As cities invest in social housing they choose between different options such as large estates vs scattered sites. Evidence on which decisions align with better neighbourhood health can inform planners & public health departments. It can also unveil structural problems which deserve attention. I expect that the results will be nuanced so I will have to take extra special care to be clear about what can be taken from the data and what cannot.

## 2. Background & Related Work

### 2.1 Housing as a determinant of health
There is a general understanding that housing affects health through several pathways. These include physical conditions (pests, overcrowding, mould, etc.), affordability (rent, less money left for food, etc.), stability (eviction, lease terms, etc.), and the neighbourhood (safety, services, walkability, etc.) [1, 3]. Krieger & Higgins [1] frame housing as a public health issue. Reviews of housing-improvement interventions found that improvements targeted at people in poor housing improve health, particularly mental & respiratory health. However, there is often weak evidence for these claims [2, 3].

**How this shapes my study:** These pathways point in opposite directions for the impacts of public housing. Affordability & stability afforded by public housing should help health outcomes whilst the poor physical condition & concentration of poverty should hurt health outcomes. The evidence suggests that there could be a net null impact but this balance would likely vary from city to city. This is why I made RQ3 ask about the type of environment instead of treating public housing as a monolith.

### 2.2 Evidence specific to public / assisted housing
- **United States:** In Fenelon et al. [4], national health survey data is linked to HUD records. They compared current public housing residents with people who would enter assistance within two years. Current public housing residents had lower odds of fair/poor self-rated health and psychological distress. This evidence suggests that public housing may be protective when isolating for the similar demographics. In contrast, cross-sectional studies find that children in NYC public housing have the highest asthma prevalence of any housing type in NYC [5].
- **Moving to Opportunity (MTO):** In a randomised experiment, public housing families who were offered vouchers to move to low-poverty neighbourhoods had about 20% lower rates of extreme obesity & diabetes [6]. This suggests that neighbourhood context around public housing matters for health, separate from the housing unit itself.
- **UK residualisation:** Over time social housing in the UK has become increasingly concentrated among lower income households ("residualisation") [7, 8]. This change in income level would impact health outcomes.

**How this shapes my study:** Together these results lead to my revised hypothesis in 1.3. I expect a raw negative correlation but that doesn't mean public housing worsens health. I need to control for demographics & income (section 4.3).

### 2.3 Mixed tenure, concentration and the Dutch model
Mixing tenures is a policy goal in all three cities. The hope is that it reduces the harms of concentrated poverty. However, systematic reviews of housing in the UK found that the benefits of tenure mix are unclear and often overstated. Reviews tend to report the positive effects even when the primary studies showed a weak relationship [9]. Dutch and Swedish researchers also found that housing mix doesn't automatically produce social mix or better outcomes for low income families [10].

**How this shapes my study:** RQ3 tests concentration versus dispersion directly with a measurable spatial indicator (Section 4.3), and it does not assume that the Amsterdam model is better. H3 is stated in a form that the data could reject.

### 2.4 Active transport and the built environment
Walkable, dense, well-connected neighbourhoods are associated with more physical activity. A 14-city study found that residents of the most activity-friendly neighbourhoods were active for 68-89 minutes more per week [11]. The Netherlands' high cycling rates are also linked to dedicated well-connected infrastructure [12]. This is the basis for my theory that Amsterdam style social housing would show better physical activity outcomes.

**How this shapes my study:** bike-network density is treated as a *moderator* (does it change the housing–activity association?), mainly for the physical-activity outcome. Because it adds scope, it remains a stretch goal.

### 2.5 Methodological literature: small-area estimates, ecological data, spatial dependence
- **Model-based health estimates.** The NYC (CDC PLACES) and Amsterdam (RIVM) health figures are modelled from survey data and demographics, not measured directly [13–15]. This means they may hide any housing effect beyond demographics. To address this, I use London's direct Census health data as the main outcome for change over time, cross-check NYC against the directly surveyed Community Health Survey [16], and check whether the RIVM model already uses housing tenure as an input.
- **Ecological fallacy.** Associations between area-level rates do not show how individuals are affected [18]. All conclusions will be stated at the neighbourhood level.
- **Modifiable areal unit problem (MAUP).** Results can change with the size and shape of the units [19]. Since the three cities use units of different sizes, I will compare *within-city standardised* associations, and I will test sensitivity by aggregating to coarser units (Section 4.4).
- **Spatial autocorrelation.** Neighbouring areas resemble each other, which violates OLS independence and overstates precision. Moran's I and spatial regression models handle this [20, 21].
- **Self-rated health as an outcome.** Self-rated general health is a strong independent predictor of mortality [22]. This justifies using it as the headline outcome that all three cities measure.

### Ethical considerations
- **Ecological fallacy:** all the data are neighbourhood aggregates, so I can't make claims about individuals. Results will be worded as "neighbourhoods with more social housing…", never "public-housing residents…".
- **Stigma:** public-housing communities are often portrayed negatively, and the Raw Hypothesis predicts an unflattering raw association. I will always show unadjusted and adjusted results side by side. I will frame the findings around structural factors (allocation by need, funding, maintenance, design), not residents, and avoid labels like "worst neighbourhoods" on maps.
- **Modelled data presented as fact:** PLACES and RIVM figures are estimates. Figures will say so, and I will not rank individual neighbourhoods.
- **Privacy:** all data are public and aggregated. No individual-level data are used. Census figures are already disclosure-controlled by ONS.
- **Western/high-income bias:** three wealthy cities don't represent housing policy globally, and I will say so when discussing how far the results generalise.

## 4. Methodology

### 4.1 Overall design and reasoning
The design has **three layers**, one per research question. Each one builds on the previous:

1. **Within-city cross-sectional regression (RQ1).** This is the core analysis. Running models *within* each city avoids comparing incompatible health measures across cities, and it gives hundreds to thousands of observations per city instead of n = 3.
2. **Within-neighbourhood change (RQ2).** Comparing each neighbourhood with itself over time (first differences) removes stable unmeasured differences between neighbourhoods, such as historical reputation or built form. These are a major source of confounding in RQ1.
3. **Effect heterogeneity (RQ3).** This tests whether RQ1's association differs by spatial concentration (and as a stretch goal, cycling infrastructure within cities) and then compares the standardised associations across the three cities.

Tools: Python (`pandas`, `geopandas`, `statsmodels`, PySAL's `libpysal`/`esda`/`spreg`, `matplotlib`/`seaborn`), the `cbsodata` package and the Census/Nomis APIs. All steps will be in the notebook so that the analysis is reproducible from raw downloads.

### 4.2 Data preparation and harmonisation
1. **Acquire** each dataset programmatically where an API exists (CBS OData, Census API, Nomis), or by documented download otherwise. Save raw files unchanged in `data/raw/`.
2. **Geography.**
   - *NYC:* aggregate HUD public-housing units to tracts and divide by ACS total housing units. Check which tract vintage (2010 vs 2020) each PLACES release uses [13], and convert to one vintage with the Census Bureau's tract relationship files (area- or population-weighted).
   - *Amsterdam:* CBS boundaries change most years. Use the RIVM 2024-classification release, which re-estimates earlier waves on 2024 boundaries, and match CBS *Kerncijfers* by code. Where codes don't match, aggregate up to *wijk*.
   - *London:* use the ONS 2011→2021 MSOA lookup. Most MSOAs are unchanged. Merges and splits are aggregated to a common "best-fit" unit.
3. **Exclusions.** Drop units with very small populations (e.g. < 500 residents: parks, industrial areas, Schiphol-type zones) and units where the exposure is suppressed. Report how many are dropped and their share of population.
4. **Coding.** Outcomes are represented as % with worse health. Social housing share is shown as a percentage. Population density is shown as a log. Within each city, continuous variables are standardised so that a SD change in one axis correlates with a SD change in the other, allowing for comparison between cities.

### 4.3 How each research question is answered

**RQ1: Cross-sectional association.** Within each city, I will regress each health outcome on social-housing share in nested steps: unadjusted (M0), adding demographics (M1), adding income (M2) and adding district fixed effects (M3). Models are population-weighted with robust standard errors. Comparing the share coefficient between M0 and M2 tests the Raw and Adjusted Hypotheses. I will check the residuals for spatial clustering (Moran's I) and switch to a spatial regression model if needed. For NYC, I will repeat the analysis with the directly surveyed Community Health Survey as a cross-check.

**RQ2: Change over time.** I will compare each neighbourhood with itself over time: the change in health against the change in social-housing share, controlling for changes in demographics and income. London (2011 → 2021 Census) is the primary case, Amsterdam (2016 → 2022) is secondary, and NYC is exploratory because PLACES methods change between releases. I will separate the loss of social homes from new private building that dilutes the share, and note that COVID-19 may affect the 2020–2022 data.

**RQ3: Type of environment.** I will classify high-share neighbourhoods as concentrated (surrounded by other high-share areas, using Local Moran's I) or dispersed, and test whether the RQ1 association differs between them. As a stretch goal, I will test whether cycling-network density changes the link between social housing and physical inactivity. Finally, I will compare the standardised associations across the three cities in a forest plot.

### 4.4 Robustness and sensitivity checks
| Check | What it addresses |

| PLACES vs Community Health Survey (NYC); RIVM predictor check (Amsterdam) | Modelled-outcome artefacts (Section 2.5) |

### 4.5 Threats to validity (summary)
- **Selection / reverse causation:** social housing is allocated by need, so poor health can *cause* a high social-housing share (e.g. disability priority in UK allocations). Controls reduce but cannot remove this, so no causal claims will be made.
- **Unmeasured confounding:** e.g. housing *quality* (age of buildings, repair backlog), which is not available consistently across cities. NYCHA completion year is a partial proxy in NYC.
- **Measurement:** modelled outcomes (PLACES, RIVM), differing definitions across countries, and Census disclosure control (small perturbations, negligible at MSOA scale).
- **Ecological inference:** neighbourhood-level only [18].
- **Multiple comparisons:** Because I am making multiple comparisons, takeaways across different comparisons may not be comparable. So, rather than counting how many results pass a significance test, I will focus on how large each relationship is and how certain I can be about it.

### 4.6 How findings will be presented
The final report (A7) will be organised around the three research questions, with one main chart for each:

- **Maps:** side-by-side maps of each city showing where social housing is and where health is worse, to give context.
- **RQ1:** a chart showing how the link between social housing and health changes as I add income and demographic controls.
- **RQ2:** a scatterplot of the change in social housing against the change in health in each London neighbourhood.
- **RQ3:** a chart comparing results for concentrated vs dispersed social housing, and a side-by-side comparison of the three cities.

All charts will use colour-blind-friendly colours and neutral labels. Charts using modelled data will say so in the caption, and raw results will always be shown next to the adjusted ones. Full regression tables will go in an appendix.

## 5. Unknowns & Dependencies

- **Different-sized areas:** London's data for obesity, exercise and mental health only comes at the borough level (large areas), so for London these will be used as background only. General health is available for small neighbourhoods.
- **Changing boundaries:** neighbourhood boundaries are redrawn over time in all three cities. I will use official lookup tables to match old and new areas, but some neighbourhoods may have to be dropped.
- **What counts as "public housing":** Dutch housing associations, NYC public housing and UK social rent are not the same thing, which could affect how the cities compare.
- **COVID-19:** some of the health surveys took place during the pandemic, which may make health look different than usual. I will note this on my charts.
- **Estimated health data:** the NYC and Amsterdam health figures are estimates rather than direct measurements, so they may not pick up the effect of housing. I will cross-check them against direct survey data where possible.
- **Data access & language:** some Dutch data is only documented in Dutch, and some data websites may change or limit downloads.
- **Cycling data:** comparing bike-lane data across all three cities would add a lot of work, so this is a stretch goal.
- **Time:** cleaning and joining data from three cities in two languages is the biggest risk to finishing on time. If needed, I will narrow the study to a single year (around 2020).
