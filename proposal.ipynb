# Public Housing and Public Health: A Three-City Comparison
**HCDE 410 Final Project Proposal**

Cities: **Amsterdam**, **New York City**, **London**. Common study window: ~ **2016–2022**.

## 1. Motivation & Problem

Housing is one of the strongest social determinants of health. Affordable, stable, and well-maintained housing can lower stress, reduce exposure to environmental hazards, and free up more household income. Cities take very different approaches to housing:

- **Amsterdam**: a large share of homes are owned by non-profit *woningcorporaties* (housing associations). They are mixed into ordinary neighbourhoods, which are dense, walkable, and built around cycling.
- **New York City**: public housing is concentrated in large NYCHA tower developments, many of them ageing and in need of repairs.
- **London**: council and housing-association homes, ranging from post-war estates to recent mixed-tenure schemes.

This contrast makes it possible to ask not only *whether* more public housing goes with better health, but *what kind* of public housing does.

**Research questions**
1. Within and across these three cities, do neighbourhoods with a larger share of public/social housing show different health outcomes (self-rated health, obesity, physical inactivity, mental health), after accounting for income and other neighbourhood characteristics?
2. When the public housing share changes over time (new construction, demolition, privatisation), do health indicators move with it?
3. Does the *type* of public housing environment matter? For example, Amsterdam's cycling-oriented, mixed-tenure model versus NYC's concentrated tower estates, measured by indicators such as physical activity and bike-network density.

**Hypothesis:** areas with a greater share of public housing will have better health outcomes, and the effect will be strongest where public housing is integrated with active-transport infrastructure.

**Why it's useful:** cities are investing heavily in social housing right now. Evidence on how housing models can be changed to improve health outcomes could inform city planners and public-health departments. I expect that the results will have more nuance and that there could be confounding variables which affect results.

### Why these fit
- **Overlap:** all three cities have both housing and health data for roughly 2016–2022 (Amsterdam health waves in 2016, 2020 and 2022, whilst PLACES & Fingertips are annually).
- **Comparable indicators:** each source has some version of self-rated general health, obesity, physical (in)activity and mental health. Definitions do differ between datasets, so I will compare within-city relationships rather than raw levels across cities.
- **Controls for confounders:** CBS, the ACS (via PLACES/HUD) and GLA data include income, age and population density, so I can control for the fact that public housing tenants are usually lower-income.

### Ethical considerations
- **Ecological fallacy:** all the data are neighbourhood aggregates so I can't make any claims on an individual level
- **Stigma:** public housing communities are often portrayed negatively. I will frame findings around structural factors (funding, maintenance, design), not the residents to reduce this bias
- **Privacy:** all the data are public and aggregated. No individual-level data will be used
- **Western/high-income bias:** three wealthy cities don't represent housing policy globally. I will say so when discussing how far the results generalise

## 3. Unknowns & Dependencies

- **Geographic mismatch:** London data are mostly at the borough level (33 units), while Amsterdam and NYC are at the neighbourhood level. I may need to aggregate NYC and Amsterdam up, or look for finer London data
- **Changing boundaries:** Dutch neighbourhood boundaries and US census tracts (2010 to 2020) change over time. I need to investigate if there are effective ways to account for this
- **Definitions of "public housing":** Dutch *corporatiewoningen*, US public housing (NYCHA), and UK social rent are not equivalent. I need to investigate how this could affect results
- **Survey disruptions:** the 2020 and 2022 health waves were affected by COVID-19, which may distort trends. I will add a note to my visualisations
- **Data access and language:** CBS/RIVM metadata are partly in Dutch, and some APIs (Fingertips, CBS OData) may change or rate-limit
- **Bike-infrastructure comparison:** comparable cycling-network data for NYC (NYC DOT bike routes) and London (TfL Cycling Infrastructure Database) exist, but they would add scope. This part of the analysis is a stretch goal if time allows since it isn't central to the research question
- **Time:** cleaning and joining three cities' data in two languages is the biggest risk to finishing by the end of the quarter. If needed, I will narrow to one shared time slice (around 2020)
